```python
import pandas as pd
df=pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
#컬럼 리스트
df.columns
#컬럼 이름
df['이름']
#컬럼이름 다수:리스트 형태로
df[['이름','수학']]
#df['이름':'과학'] #슬라이싱x
df[df.columns[0:4]] #columns 사용해서 슬라이싱 적용
df.loc[:,'이름':'수학'] #loc를 사용해서 슬라이싱 가능

df[df.columns[-1]] #마지막 컬럼 출력
df[df.columns[[0,3,1]]]
df[0:3] #index에 해당되는 3개 출력
```

```python
#loc출력-index출력(이름으로 출력)
df.loc['1번']
df.loc[['1번','2번']]
df.loc['1번':'5번']
df.loc['1번','이름':'수학'] #슬라이싱 가능
df.loc['2번':'5번','이름':'수학'] #슬라이싱 가능
```

```python
#iloc출력-리스트 번호로 출력
df.iloc[0]
df.iloc[0:4]
df.iloc[[0,1,5]]
df.iloc[0:4,1:5]
```
